```python
import pandas as pd

data = {
    "Name": ["Arun", "Bala", "Charan", "Divya", "Ezhil"],
    "Department": ["CSE", "AIML", "CSE", "AIML", "CSE"],
    "Python": [80, 90, 70, 95, 85],
    "ML": [75, 85, 65, 90, 80],
    "SQL": [85, 75, 75, 95, 80]
}

df = pd.DataFrame(data)

# Select Name and ML
print(df[["Name", "ML"]])

# Students with ML > 80
print(df[df["ML"] > 80])

# AIML students
print(df[df["Department"] == "AIML"])

# Calculate average for each student
df["Average"] = df[["Python", "ML", "SQL"]].mean(axis=1)

# Student with highest average
print(df[df["Average"] == df["Average"].max()])
```

```python
import pandas as pd

df = pd.read_csv("titanic.csv")

print(df.head())
```

```python
print("Shape of dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df = df.drop("Cabin", axis=1)
```

```python
cleaned_df = df.copy()
```

```python
cleaned_df["Age"] = cleaned_df["Age"].fillna(cleaned_df["Age"].median())

cleaned_df["Embarked"] = cleaned_df["Embarked"].fillna(cleaned_df["Embarked"].mode()[0])
```

```python
print(cleaned_df.isnull().sum())
```

```python
print("Number of duplicate rows:", cleaned_df.duplicated().sum())
```

```python
print(cleaned_df["Survived"].value_counts())
```

```python
survival_rate = cleaned_df["Survived"].mean() * 100

print("Survival Rate:", survival_rate, "%")
```

```python
import matplotlib.pyplot as plt

survival_counts = cleaned_df["Survived"].value_counts()

survival_counts.plot(kind="bar")

plt.title("Titanic Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
gender_survival = cleaned_df.groupby("Sex")["Survived"].mean() * 100

print(gender_survival)
```

```python
gender_survival.plot(kind="bar")

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")

plt.show()
```

```python
## EDA Finding

The analysis shows that the survival rate differed significantly between male and female passengers. Female passengers had a higher survival rate than male passengers in the Titanic dataset.
```

```python
## EDA Findings

### Survival Analysis
The Titanic dataset contains 891 passenger records. The overall survival rate was approximately 38%.

### Gender Analysis
The analysis shows that female passengers had a significantly higher survival rate than male passengers.

### Data Cleaning
Missing values in the Age column were filled using the median. Missing values in the Embarked column were filled using the mode. The Cabin column was removed because it contained a large number of missing values.

No duplicate rows were found in the cleaned dataset.
```
